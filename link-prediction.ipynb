# Comparison of Graph Link Prediction Methods

Compare classical similarity scores, embeddings and GNNs on the same held-out links. The default example uses offline karate-club data and does not download datasets.


```python
from datetime import datetime, timezone
from pathlib import Path
from IPython.display import Image, display
from linkpred.experiment import RunConfig, run_experiment

output = Path('reports') / ('notebook-' + datetime.now(timezone.utc).strftime('%Y%m%dT%H%M%S%fZ'))
config = RunConfig(dataset='karate', dimensions=8, epochs=2, walk_length=5, num_walks=3)
report = run_experiment(config, methods=('jaccard', 'adamic-adar', 'hope'), output=output)
```

```python
for result in report['results']:
    print(result['method'], result['status'], result.get('test', {}))
print('Shared split:', report['metadata']['split_id'])
display(Image(filename=str(output / 'scores.png')))
display(Image(filename=str(output / 'runtime.png')))
```
